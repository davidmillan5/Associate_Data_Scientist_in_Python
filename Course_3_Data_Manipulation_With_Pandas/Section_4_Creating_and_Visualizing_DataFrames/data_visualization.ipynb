```python
# Import matplotlib.pyplot with alias plt
import matplotlib.pyplot as plt
```

```python
# Import pandas with alias pd
import pandas as pd
```

```python
avocados = pd.read_csv('avocados.csv')
```

```python
# Look at the first few rows of data
print(avocados.head())
```

```python
unique_size = avocados['size'].unique()
print(unique_size)
```

```python
nb_sold_by_size = (avocados.groupby('size', as_index=False)['nb_sold'].sum())
print(nb_sold_by_size)
```

```python
# Get the total number of avocados sold of each size
nb_sold_by_size = (avocados.groupby('size', as_index=False)['nb_sold'].sum())
```

```python
nb_sold_by_size.plot(kind='bar')
```

```python
nb_sold_by_size = avocados.groupby('size')['nb_sold'].sum()
print(nb_sold_by_size)
```

```python
nb_sold_by_size.plot(kind='bar')
```

```python
nb_sold_by_date = avocados.groupby('date')['nb_sold'].sum()
print(nb_sold_by_date)
```

```python
nb_sold_by_date.plot(
    x='date',
    y='nb_sold',
    kind='line',
    rot= 45
)
plt.show()
```

```python
# Scatter plot of avg_price vs. nb_sold with title
avocados.plot(x='nb_sold', y='avg_price', kind='scatter', title="Number of avocados sold vs. average price")
```

```python
# Show the plot
plt.show()
```

```python
unique_type = avocados['type'].unique()
print(unique_type)
```

```python
# Histogram of conventional avg_price
avocados[avocados['type'] == 'conventional']['avg_price'].hist()
```

```python
# Histogram of organic avg_price
avocados[avocados['type'] == 'organic']['avg_price'].hist()
```

```python
# Add a legend
plt.legend(['conventional', 'organic'])
```

```python
# Modify histogram transparency to 0.5
avocados[avocados["type"] == "conventional"]["avg_price"].hist(alpha = 0.5)

# Modify histogram transparency to 0.5
avocados[avocados["type"] == "organic"]["avg_price"].hist(alpha = 0.5)

# Add a legend
plt.legend(["conventional", "organic"])

# Show the plot
plt.show()
```

```python
# Modify bins to 20
avocados[avocados["type"] == "conventional"]["avg_price"].hist(alpha=0.5, bins=20)

# Modify bins to 20
avocados[avocados["type"] == "organic"]["avg_price"].hist(alpha=0.5, bins=20)

# Add a legend
plt.legend(["conventional", "organic"])

# Show the plot
plt.show()
```
